```python
# 1. Lists & In-Place Slicing
sales = [45000, 52000, 61000, 72000]
sales.append(85000)
print('Sales List:', sales)
print('Top 2 Months:', sales[:2])
```

```python
# 2. Sets (Unique items & Venn Operations)
jan_buyers = {'Aarav', 'Neha', 'Rohan', 'Priya'}
feb_buyers = {'Priya', 'Vikram', 'Sneha', 'Aarav'}

print('Repeat Customers in Both Months:', jan_buyers.intersection(feb_buyers))
print('Total Unique Customers:', jan_buyers.union(feb_buyers))
print('Churned (Jan Only):', jan_buyers.difference(feb_buyers))
```

```python
# 3. Dictionaries & Comprehensions
prices = {'Laptop': 65000, 'Mouse': 1200, 'Monitor': 18000, 'Keyboard': 2500}
# Apply 10% discount using dictionary comprehension
discounted = {item: round(p * 0.90, 2) for item, p in prices.items()}
print('Discounted Catalog:', discounted)
```
