**Objective:** Run SQL Queries directly in your browser without installing MySQL server.

```python
import sqlite3
import pandas as pd

# Create in-memory SQLite database
conn = sqlite3.connect(':memory:')

# Create table
conn.execute('''
CREATE TABLE customers (
    id INTEGER PRIMARY KEY,
    name TEXT,
    city TEXT,
    spend REAL
);
''')

# Insert sample data
conn.executemany('INSERT INTO customers VALUES (?, ?, ?, ?)', [
    (1, 'Aarav Sharma', 'Mumbai', 85000),
    (2, 'Neha Verma', 'Delhi', 45000),
    (3, 'Rohan Gupta', 'Bangalore', 62000),
    (4, 'Priya Nair', 'Pune', 120000)
])
conn.commit()

# Query with SQL into Pandas DataFrame
query = 'SELECT * FROM customers WHERE spend > 50000 ORDER BY spend DESC;'
df_sql = pd.read_sql_query(query, conn)
print('High-Spend Customers Query Result:')
display(df_sql)
```
