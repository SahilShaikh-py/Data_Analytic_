```python
import numpy as np
import pandas as pd

# NumPy Vectorized Operations
sales_q1 = np.array([12000, 15000, 19000])
sales_q2 = np.array([14000, 16500, 22000])
total_h1 = sales_q1 + sales_q2
print('H1 Combined Sales:', total_h1)
```

```python
# Creating a Pandas DataFrame
data = {
    'Department': ['IT', 'HR', 'IT', 'Finance', 'Marketing', 'IT', 'Finance'],
    'Salary': [65000, 52000, 95000, 80000, 72000, 68000, 84000],
    'Experience': [3, 5, 11, 7, 15, 4, 9]
}
df = pd.DataFrame(data)
print(df.head())

# Department Summary Aggregation
dept_summary = df.groupby('Department')['Salary'].agg(['mean', 'count', 'max'])
print('\nDepartment Salary Summary:')
print(dept_summary)
```
