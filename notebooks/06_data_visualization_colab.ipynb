```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [45000, 52000, 61000, 58000, 74000, 89000]

plt.figure(figsize=(8, 4))
plt.plot(months, sales, marker='o', color='#1E88E5', linewidth=2.5)
plt.title('Monthly Sales Performance (2026)', fontsize=14, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Revenue (INR)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
