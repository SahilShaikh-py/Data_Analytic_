**Objective:** Handle Missing Values, Outliers, Correlation Matrix, and Category Slicing.

```python
import pandas as pd
import numpy as np

raw_dataset = {
    'Customer': ['Aarav', 'Neha', 'Rohan', 'Priya', 'Aarav'],
    'Age': [28, np.nan, 35, 29, 28],
    'Sales': [15000, 22000, np.nan, 45000, 15000]
}
df = pd.DataFrame(raw_dataset)
print('Raw with Duplicates & Nulls:')
print(df)

# Deduplicate
df = df.drop_duplicates()
# Impute missing values
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Sales'] = df['Sales'].fillna(df['Sales'].mean())

print('\nCleaned DataFrame:')
print(df)
```
