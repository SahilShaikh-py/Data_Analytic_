## 1. Variables & Primitive Types
Python dynamically infers variable types at runtime.

```python
student_name = 'Sahil Shaikh'
student_age = 22
course_gpa = 3.85
is_enrolled = True

print(f'Name: {student_name} ({type(student_name).__name__})')
print(f'Age: {student_age} ({type(student_age).__name__})')
print(f'GPA: {course_gpa} ({type(course_gpa).__name__})')
print(f'Active: {is_enrolled} ({type(is_enrolled).__name__})')
```

## 2. Type Casting & f-Strings
Convert string inputs from external files into numbers for business calculation.

```python
raw_sales_input = '1500'
tax_percentage = '0.18'

sales_amount = float(raw_sales_input)
tax_rate = float(tax_percentage)
grand_total = sales_amount * (1 + tax_rate)

print(f'Sales: Rs. {sales_amount:,.2f}')
print(f'Grand Total with Tax: Rs. {grand_total:,.2f}')
```

## 3. Conditional Statements (if-elif-else)
Categorize customers into discount loyalty tiers based on order volume.

```python
order_total = 8500

if order_total >= 10000:
    tier = 'Platinum'
    discount = 0.20
elif order_total >= 5000:
    tier = 'Gold'
    discount = 0.12
else:
    tier = 'Standard'
    discount = 0.0

savings = order_total * discount
print(f'Tier: {tier} | Discount: {discount*100}% | Savings: Rs. {savings:,.2f}')
```

## 4. Loops & Iteration
Iterating over weekly revenue figures.

```python
daily_revenue = [12000, 15400, 18900, 14200, 22500, 28000, 31000]
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

for day, rev in zip(days, daily_revenue):
    print(f'{day}: Rs. {rev:,}')

print(f'Total Weekly Revenue: Rs. {sum(daily_revenue):,}')
```
